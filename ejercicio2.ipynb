```python
import numpy as np
from scipy.integrate import simpson, trapezoid

# 1. Cargar datos desde el CSV usando NumPy
data = np.loadtxt('datos_sensor.csv', delimiter=',')
x = data[:, 0]
y = data[:, 1]

N = len(x)
h = x[1] - x[0]

# ---------------------------------------------------------
# A. DIFERENCIACIÓN NUMÉRICA dy/dx
# ---------------------------------------------------------
dydx = np.zeros(N)

# Extremo inicial: Adelante O(h)
dydx[0] = (y[1] - y[0]) / h

# Puntos intermedios: Centrada O(h^2)
dydx[1:-1] = (y[2:] - y[:-2]) / (2 * h)

# Extremo final: Atrás O(h)
dydx[-1] = (y[-1] - y[-2]) / h

# ---------------------------------------------------------
# B. INTEGRACIÓN NUMÉRICA (3 MÉTODOS)
# ---------------------------------------------------------
# 1. Rectángulo
I_rect = h * np.sum(y[:-1])

# 2. Trapecio (Manual y SciPy)
I_trap_manual = (h / 2) * (y[0] + 2 * np.sum(y[1:-1]) + y[-1])
I_trap_scipy = trapezoid(y, x)

# 3. Simpson (SciPy)
I_simp_scipy = simpson(y, x=x)

# ---------------------------------------------------------
# C. IMPRESIÓN DE RESULTADOS
# ---------------------------------------------------------
print("=========================================================")
print("  EJERCICIO 2: PROCESAMIENTO DE DATOS DISCRETOS (PYTHON)")
print("=========================================================")
print(f"  Total de mediciones (N) : {N}")
print(f"  Paso de muestreo (h)    : {h:.4f}")
print("---------------------------------------------------------")
print(f"  Integral Rectángulo       : {I_rect:.8f}")
print(f"  Integral Trapecio (Manual): {I_trap_manual:.8f}")
print(f"  Integral Trapecio (SciPy) : {I_trap_scipy:.8f}")
print(f"  Integral Simpson (SciPy)  : {I_simp_scipy:.8f}")
print("---------------------------------------------------------")
print(f"  Primera derivada dy/dx(0) : {dydx[0]:.6f}")
print(f"  Derivada central dy/dx(1) : {dydx[1]:.6f}")
print(f"  Última derivada dy/dx(end): {dydx[-1]:.6f}")
print("=========================================================")
```
