```python
import numpy as np
from scipy.integrate import quad, simpson, trapezoid

# Definición de la función continua
def f(x):
    return 2.0 + 0.35 * np.sin(0.7 * x) + 0.15 * np.cos(2.1 * x) + 0.03 * x

a, b = 0.0, 9.8
N = 100
x = np.linspace(a, b, N + 1)
y = f(x)
h = (b - a) / N

# 1. Métodos Newton-Cotes
I_rect = h * np.sum(y[:-1])
I_trap = trapezoid(y, x)
I_simp = simpson(y, x=x)

# 2. Referencia adaptativa
I_quad, err_quad = quad(f, a, b)

print("=========================================================")
print(f"  EJERCICIO 1: FUNCIÓN CONTINUA (PYTHON - N={N})")
print("=========================================================")
print(f"  Integral Rectángulo       : {I_rect:.8f} | Err: {abs(I_rect - I_quad):.2e}")
print(f"  Integral Trapecio         : {I_trap:.8f} | Err: {abs(I_trap - I_quad):.2e}")
print(f"  Integral Simpson 1/3      : {I_simp:.8f} | Err: {abs(I_simp - I_quad):.2e}")
print(f"  Integral SciPy quad (Ref) : {I_quad:.8f}")
print("=========================================================")
```
